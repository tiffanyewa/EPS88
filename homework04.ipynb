## Exercise 3 - Evaluating the model using $R^{2}$

We'd also like to know who well this model fits our data (i.e. how correlated the data are). The $R^{2}$ correlation coefficient can be helpful in this regard. $R^{2}$ is zero for uncorrelated data, and 1 for perfectly linear data (so no misfit between the model line and data). 

**(5 points)** Let's calculate the $R^{2}$ value for our model.
Recall that the $R^{2}$ value is calculated as follows:

$$R^{2} = 1 - \frac{\sum_{i=1}^{n} (y_{i} - \hat{y}_{i})^{2}}{\sum_{i=1}^{n} (y_{i} - \bar{y})^{2}}$$

where $y_{i}$ is the actual data, $\hat{y}_{i}$ is the prediction, and $\bar{y}$ is the mean of the actual data.

```python
## Calculate the R^2 value for the model
from sklearn.metrics import r2_score
R_squared=r2_score(P566_GPS_data[[' North (mm)']],y_pred)

print(f'R^2= {R_squared:.4f}')
```

### Defining a function

When you may be doing a calculation more than once it is a good idea to define a function. Let's define a function that will take an east magnitude and a north magnitude and return a direction between 0 and 360.

```python
def GPS_direction(east_magnitude, north_magnitude):
    direction_rad = np.arctan2(east_magnitude, north_magnitude)
    direction = np.rad2deg(direction_rad) % 360
    return direction
```

Are the slopes of the two periods siginficantly different from each other? Is there a change in the land subsidence rate through time?

Remember (from lectures), the standard error of the slope for linear regression is:

$$ SE_{s} = \sqrt{\sigma^2 / \sum{(x_{i} - \bar{x})}^2} $$

where $\sigma^2 = \sum{(y_i - \hat{y}_i)^2}/(n-p) $ and $n$ is the number of observations, $p$ is the number of predictors. Together $n - p$ equals the 'degrees of freedom' $dof$.

The standard error of the intercept can then be given by:

$$ SE_{I} = \sqrt{\sigma^2(1/n + \bar{x}^2/\sum{(x_i - \bar{x})**2})} $$

Remember $\bar{x}$ indicates the mean value of $x$ and $\hat{y}_i$ indicates the predicted value of $y$ at position $i$.

```python
X = P566_GPS_18_23['days']
n = len(X)
p = len(model_18_23.coef_[0]) + 1
dof = n - p
y = P566_GPS_18_23[' Vertical (mm)']
y_pred = model_18_23.predict(X.values.reshape(-1,1))

sigma = np.sum((y - y_pred.flatten())**2) /dof

se_slope = np.sqrt(sigma/np.sum((X - X.mean())**2))
se_inter = np.sqrt(sigma * (1/n + (X.mean()**2)/np.sum((X - X.mean())**2)))

print('2018-2023:')
print(f"Standard error of slope: {se_slope:.4f}")
print(f"Standard error of intercept: {se_inter:.4f}")

X1 = P566_GPS_06_11['days']
n1 = len(X1)
p1 = len(model_06_11.coef_[0]) + 1
dof1 = n - p
y1 = P566_GPS_06_11[' Vertical (mm)']
y1_pred = model_06_11.predict(X1.values.reshape(-1,1))

sigma1 = np.sum((y1 - y1_pred.flatten())**2) /dof

se_slope1 = np.sqrt(sigma1/np.sum((X1 - X1.mean())**2))
se_inter1 = np.sqrt(sigma1 * (1/n1 + (X1.mean()**2)/np.sum((X1 - X1.mean())**2)))

print('\n')
print('2006-2011:')
print(f"Standard error of slope: {se_slope1:.4f}")
print(f"Standard error of intercept: {se_inter1:.4f}")
```
