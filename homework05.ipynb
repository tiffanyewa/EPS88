# Homework 5 - Earthquake Probability

In class we used the observed rate of earthquakes in the Bay Area to estimate the Gutenberg-Richter statistic. In class we did it over a 6 year period but we could recalculate this to give us the annual rate of earthquakes within different magnitude 'bins' (as before we'll use bins of 0.1 Magnitude units). As such, we can then say over some period of time, how many earthquakes above a certain magnitude should we expect to see.

The occurence of earthquakes if often thought of as a Poisson process, events occur randomly in time and the average rate of recurrance is known. For an individual fault strand/segment this might not be true (often we consider that there is some time required for the necessary strain energy to rebuild on the fault), but for regions of seismic activity more generally it is a good approximation. When estimating earthquake probabilities it is common to consider ${\Delta}time=30 yrs$. Why is that?

The Gutenberg-Richter relationship for the entire San Franciso Bay Area (determined using a longer earthquake record than we used in class):

$$ log_{10}(N)= 3.68 - 0.82\times M $$

**In the following notebook your assignment is to estimate - using the Poisson distribution - the probability of one or more earthquakes with magnitude > 4, 5, 6, and 7 in the next 30 years.**

```python
# Possion distributions of the probability of one or more earthquakes with magnitude higher then 4, 5, 6, and 7 in the next 30 years

import numpy as np
from scipy.stats import poisson
import matplotlib.pyplot as plt

# the Gutenberg-Richter parameters for the entire San Fransuco Bay Area
A = 3.68
b = 0.82


# Magnitudes to look at 
Mag = [4, 5, 6, 7]

# Period 
time_30 = 30

for M in Mag:
    # the annual rate from Gutenberg-Richter
    log1= 10**(A - b * M)
    
    # expected number of earthquakes over the time period = 30
    lambda_t = log1 * time_30
    
    # the probability of one or more earthquakes
    prob_one_over = 1 - poisson.pmf(0, lambda_t)
    
    # the robability of more than 2 earthquakes
    prob_over_2 = 1 - poisson.cdf(2, lambda_t)
    
    print(f'M >= {M}: annual rate = {log1:.2f}, yrs = {lambda_t:.2f}')
    print(f' One or more = {prob_one_over:.2f} ({prob_one_over*100:.2f}%)')
    print(f' Two or more = {prob_over_2:.2f} ({prob_over_2*100:.2f}%)')
    print()
```

```python
# Calculate the propbiliblites if the time period is 10 years instead of 30 years

# Period 
time_10 = 10

for M in Mag:
    # the annual rate from Gutenberg-Richter
    log1= 10**(A - b * M)
    
    # expected number of earthquakes over the time period = 30
    lambda_t = log1 * time_10 
    
    # the probability of one or more earthquakes
    prob_one_over = 1 - poisson.pmf(0, lambda_t)
    
    # the robability of more than 2 earthquakes
    prob_over_2 = 1 - poisson.cdf(2, lambda_t)
    
    print(f'M >= {M}: annual rate = {log1:.2f}, yrs = {lambda_t:.2f}')
    print(f' One or more = {prob_one_over:.2f} ({prob_one_over*100:.2f}%)')
    print(f' Two or more = {prob_over_2:.2f} ({prob_over_2*100:.2f}%)')
    print()
```
